```python
import json

filmes = []

while True:
    nome_filme = input("Digite o nome do filme/série ").strip()
    diretor = input("Diretor (ou outra informação textual): ").strip()

    while True:
        try:
            avaliacao = float(input("Avaliação (de 0 a 10): "))
            if 0 <= avaliacao <= 10:
                break
            else:
                print("Avaliação inválida. Por favor, digite um valor entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um valor numérico válido.")

    data = input("Data de lançamento (xxxx):").strip()

    while True:
        try:
            horas = int(input("Duração (em horas): "))
            minutos = int(input("Duração (em minutos): "))
            if horas >= 0 and minutos >= 0:
                break
            else:
                print("Duração inválida. Por favor, digite valores não negativos.")
        except ValueError:
            print("Entrada inválida. Por favor, digite valores numéricos válidos.")

    # validacao do booleano
    while True:
        streaming = input("Streaming (s/n): ").strip().lower()
        if streaming in ["s", "n"]:
            disponivel_streaming = streaming == "s"
            break
        else:
            print("Digite 's' para sim ou 'n' para não.")

    # dicionario e adiconar lista
    filme = {
        "nome": nome_filme,
        "diretor": diretor,
        "avaliacao": avaliacao,
        "data": data,
        "horas": horas,
        "minutos": minutos,
        "disponivel_streaming": disponivel_streaming
    }
    filmes.append(filme)

    continuar = input("Deseja adicionar mais filmes? (s/n): ").strip().lower()
    if continuar != "s":
        break

with open("filmes.json", "w", encoding="utf-8") as f:
    json.dump(filmes, f, ensure_ascii=False, indent=4)
print("Dados salvos com sucesso!")
```

```python
def nomes_extremos_duracao(filmes):
    def minutos_totais(filme):
        h, m = filme["duracao"]
        return h * 60 + m

    if not filmes:
        return None, None

    maior = max(filmes, key=minutos_totais)
    menor = min(filmes, key=minutos_totais)

    return maior["nome"], menor["nome"]

maior_nome, menor_nome = nomes_extremos_duracao(filmes)

print(f"Filme com MAIOR duração: {maior_nome}")
print(f" Filme com MENOR duração: {menor_nome}")
```

```python
def moda_diretor(filmes):
    contagem = {}

    for filme in filmes:
        diretor = filme["diretor"]
        contagem[diretor] = contagem.get(diretor, 0) + 1

    mais_frequente = max(contagem.items(), key=lambda item: item[1])
    return mais_frequente[0], mais_frequente[1]

diretor, quantidade = moda_diretor(filmes)
print(f"O Diretor mais frequente é: {diretor} ({quantidade} filmes)")
```

```python
def media_recursiva(lista):
    def somar_e_contar(valores):
        soma = 0
        quantidade = 0
        for item in valores:
            if isinstance(item, list):
                s, q = somar_e_contar(item)
                soma += s
                quantidade += q
            else:
                soma += item
                quantidade += 1
        return soma, quantidade

    total, n = somar_e_contar(lista)
    return total / n if n > 0 else 0
```

```python
# Média alvo baseada em dados desestruturados
def ajustar_media(valores, media_alvo, fator_ajuste=0.1):
    """
    Ajusta uma lista de valores para que sua média se aproxime de uma média alvo.

    Args:
        valores: Lista de valores numéricos.
        media_alvo: A média desejada.
        fator_ajuste: Um fator para controlar a magnitude do ajuste.

    Returns:
        Uma nova lista com os valores ajustados.
    """
    if not valores:
        return []

    media_atual = sum(valores) / len(valores)
    diferenca_media = media_alvo - media_atual
    valores_ajustados = []

    for valor in valores:
        # Ajusta cada valor proporcionalmente à sua distância da média atual
        ajuste = diferenca_media * fator_ajuste * (valor / media_atual if media_atual != 0 else 0)
        valores_ajustados.append(valor + ajuste)

    return valores_ajustados

media_alvo = media_recursiva([[6.3, 2.1, [8.4, 4.2, 5.1], 9.6], 10, 4.7, 6.5])

# Lista original de avaliações
avaliacoes_alunos = [5.0, 6.0, 7.0, 4.0, 8.0]

# Ajustar para se aproximar da média alvo
avaliacoes_ajustadas = ajustar_media(avaliacoes_alunos, media_alvo)

print(f"Média alvo: {media_alvo:.2f}")
print(f"Média original: {sum(avaliacoes_alunos)/len(avaliacoes_alunos):.2f}")
print(f"Média ajustada: {sum(avaliacoes_ajustadas)/len(avaliacoes_ajustadas):.2f}")
print("Avaliações ajustadas:", avaliacoes_ajustadas)
```

```python
import csv

# 1. Função recursiva para média alvo
def media_recursiva(lista):
    def somar_e_contar(valores):
        soma = 0
        quantidade = 0
        for item in valores:
            if isinstance(item, list):
                s, q = somar_e_contar(item)
                soma += s
                quantidade += q
            else:
                soma += item
                quantidade += 1
        return soma, quantidade

    total, n = somar_e_contar(lista)
    return total / n if n > 0 else 0

# Função para ajustar as notas
def ajustar_media(avaliacoes, media_alvo):
    atual = sum(avaliacoes) / len(avaliacoes)
    diferenca = media_alvo - atual
    return [min(max(nota + diferenca, 0), 10) for nota in avaliacoes]

# Dados de entrada
avaliacoes_desestruturadas = [[6.3, 2.1, [8.4, 4.2, 5.1], 9.6], 10, 4.7, 6.5]
media_alvo = media_recursiva(avaliacoes_desestruturadas)

alunos = [
    {"nome": "Ana", "nota": 5.0},
    {"nome": "Bruno", "nota": 6.0},
    {"nome": "Carla", "nota": 7.0},
    {"nome": "Daniel", "nota": 4.0},
    {"nome": "Eduarda", "nota": 8.0}
]

# 4. Atualizar notas
notas_originais = [aluno["nota"] for aluno in alunos]
notas_ajustadas = ajustar_media(notas_originais, media_alvo)

for i, aluno in enumerate(alunos):
    aluno["nota_ajustada"] = round(notas_ajustadas[i], 2)

# 5. Exportar para CSV
with open("avaliacoes_ajustadas.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["nome", "nota_original", "nota_ajustada"])
    writer.writeheader()
    for aluno in alunos:
        writer.writerow({
            "nome": aluno["nome"],
            "nota_original": aluno["nota"],
            "nota_ajustada": aluno["nota_ajustada"]
        })

print(" Arquivo 'avaliacoes_ajustadas.csv' gerado com sucesso.")
```

```python
import csv

# Função recursiva para média alvo
def media_recursiva(lista):
    def somar_e_contar(valores):
        soma = 0
        quantidade = 0
        for item in valores:
            if isinstance(item, list):
                s, q = somar_e_contar(item)
                soma += s
                quantidade += q
            elif isinstance(item, (int, float)):
                soma += item
                quantidade += 1
            else:
                print(f"Valor ignorado: {item}")
        return soma, quantidade

    total, n = somar_e_contar(lista)
    if n == 0:
        raise ValueError("Lista não contém valores numéricos válidos.")
    return total / n

# Função para ajustar as notas
def ajustar_media(avaliacoes, media_alvo):
    if not avaliacoes:
        raise ValueError("Lista de avaliações está vazia.")
    if not all(isinstance(nota, (int, float)) for nota in avaliacoes):
        raise TypeError("Todas as avaliações devem ser números.")

    atual = sum(avaliacoes) / len(avaliacoes)
    diferenca = media_alvo - atual
    return [min(max(nota + diferenca, 0), 10) for nota in avaliacoes]

# Dados de entrada
avaliacoes_desestruturadas = [[6.3, 2.1, [8.4, 4.2, 5.1], 9.6], 10, 4.7, 6.5]
alunos = [
    {"nome": "Ana", "nota": 5.0},
    {"nome": "Bruno", "nota": 6.0},
    {"nome": "Carla", "nota": 7.0},
    {"nome": "Daniel", "nota": 4.0},
    {"nome": "Eduarda", "nota": 8.0}
]

try:
    # Calcular média alvo
    media_alvo = media_recursiva(avaliacoes_desestruturadas)

    # Extrair notas originais
    notas_originais = [aluno["nota"] for aluno in alunos]

    # Ajustar notas
    notas_ajustadas = ajustar_media(notas_originais, media_alvo)

    # Atualizar alunos
    for i, aluno in enumerate(alunos):
        aluno["nota_ajustada"] = round(notas_ajustadas[i], 2)

    # Exportar para CSV
    with open("avaliacoes_ajustadas.csv", "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["nome", "nota_original", "nota_ajustada"])
        writer.writeheader()
        for aluno in alunos:
            writer.writerow({
                "nome": aluno["nome"],
                "nota_original": aluno["nota"],
                "nota_ajustada": aluno["nota_ajustada"]
            })

    print("Arquivo 'avaliacoes_ajustadas.csv' gerado com sucesso.")

except Exception as e:
    print(f" Erro encontrado: {e}")
```
